```python
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import f1_score

from xgboost import XGBClassifier


# =====================================================
# 1. Create Models
# =====================================================

models = {
    "LR": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "KNN": KNeighborsClassifier(),

    "NB": GaussianNB(),

    "SVM": SVC(
        probability=True,
        random_state=42
    ),

    "DT": DecisionTreeClassifier(
        random_state=42
    ),

    "RF": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "XGB": XGBClassifier(
        n_estimators=100,
        random_state=42,
        eval_metric="logloss"
    )
}


# =====================================================
# 2. Train Models + Calculate Train/Test F1
# =====================================================

classification_data = []

for name, model in models.items():

    # Train model
    model.fit(X_train, y_train)

    # Train prediction
    train_pred = model.predict(X_train)

    # Test prediction
    test_pred = model.predict(X_test)

    # F1 scores
    train_f1 = f1_score(
        y_train,
        train_pred,
        zero_division=0
    )

    test_f1 = f1_score(
        y_test,
        test_pred,
        zero_division=0
    )

    # Train-Test Gap
    gap = train_f1 - test_f1

    # Bias-Variance interpretation
    if gap <= 0.05:
        fit = "GoodFit"
    else:
        fit = "Overfit"

    classification_data.append([
        name,
        train_f1,
        test_f1,
        fit
    ])


# =====================================================
# 3. Create Classification Results
# =====================================================

classification_results = pd.DataFrame(
    classification_data,
    columns=[
        "Models",
        "Train F1",
        "Test F1",
        "Bias-Variance"
    ]
)


# =====================================================
# 4. Cross Validation F1
# =====================================================

cv_data = []

for name, model in models.items():

    cv_score = cross_val_score(
        model,
        X_train,
        y_train,
        cv=5,
        scoring="f1"
    ).mean()

    cv_data.append([
        name,
        cv_score
    ])


# =====================================================
# 5. Create Cross Validation DataFrame
# =====================================================

cv_df = pd.DataFrame(
    cv_data,
    columns=[
        "Models",
        "CrossValidation"
    ]
)


# =====================================================
# 6. Merge Classification + Cross Validation
# =====================================================

final_classification_table = classification_results.merge(
    cv_df,
    on="Models",
    how="inner"
)


# =====================================================
# 7. Arrange Columns
# =====================================================

final_classification_table = final_classification_table[
    [
        "Models",
        "Train F1",
        "Test F1",
        "Bias-Variance",
        "CrossValidation"
    ]
]


# =====================================================
# 8. Round Values
# =====================================================

final_classification_table = final_classification_table.round(2)


# =====================================================
# 9. Display Final Table
# =====================================================

final_classification_table
```

```python
# ============================================
# HYPERPARAMETER TUNING + FINAL EVALUATION
# ============================================

import pandas as pd

from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

# --------------------------------------------
# 1. Check required train/test data
# --------------------------------------------

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


# ============================================
# 2. LOGISTIC REGRESSION TUNING
# ============================================

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_params = {
    "C": [0.01, 0.1, 1, 10],
    "class_weight": [None, "balanced"],
    "solver": ["liblinear", "lbfgs"]
}

lr_grid = GridSearchCV(
    lr_model,
    lr_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

lr_grid.fit(X_train, y_train)

best_lr_model = lr_grid.best_estimator_

print("\nLogistic Regression")
print("Best Parameters:", lr_grid.best_params_)
print("Best CV F1:", lr_grid.best_score_)


# ============================================
# 3. DECISION TREE TUNING
# ============================================

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_params = {
    "max_depth": [3, 5, 10, 15],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5],
    "class_weight": [None, "balanced"]
}

dt_grid = GridSearchCV(
    dt_model,
    dt_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

dt_grid.fit(X_train, y_train)

best_dt_model = dt_grid.best_estimator_

print("\nDecision Tree")
print("Best Parameters:", dt_grid.best_params_)
print("Best CV F1:", dt_grid.best_score_)


# ============================================
# 4. RANDOM FOREST TUNING
# ============================================

rf_model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

rf_params = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "class_weight": [None, "balanced"]
}

rf_grid = GridSearchCV(
    rf_model,
    rf_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

best_rf_model = rf_grid.best_estimator_

print("\nRandom Forest")
print("Best Parameters:", rf_grid.best_params_)
print("Best CV F1:", rf_grid.best_score_)


# ============================================
# 5. KNN TUNING
# ============================================

knn_model = KNeighborsClassifier()

knn_params = {
    "n_neighbors": [3, 5, 7, 9, 11],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan"]
}

knn_grid = GridSearchCV(
    knn_model,
    knn_params,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

knn_grid.fit(X_train, y_train)

best_knn_model = knn_grid.best_estimator_

print("\nKNN")
print("Best Parameters:", knn_grid.best_params_)
print("Best CV F1:", knn_grid.best_score_)


# ============================================
# 6. CREATE TUNED MODEL DICTIONARY
# ============================================

models = {
    "Logistic Regression": best_lr_model,
    "Decision Tree": best_dt_model,
    "Random Forest": best_rf_model,
    "KNN": best_knn_model
}


# ============================================
# 7. EVALUATE ALL TUNED MODELS
# ============================================

tuned_results = []

for name, model in models.items():

    y_pred = model.predict(X_test)

    # Probability for ROC-AUC and PR-AUC
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = None

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    if y_prob is not None:
        roc_auc = roc_auc_score(y_test, y_prob)
        pr_auc = average_precision_score(y_test, y_prob)
    else:
        roc_auc = None
        pr_auc = None

    tuned_results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc,
        pr_auc
    ])


# ============================================
# 8. FINAL TUNED CLASSIFICATION TABLE
# ============================================

tuned_classification_results = pd.DataFrame(
    tuned_results,
    columns=[
        "Models",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "ROC_AUC",
        "PR_AUC"
    ]
)

tuned_classification_results
```
